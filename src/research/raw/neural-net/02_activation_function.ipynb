# [Activation Functions](https://en.wikipedia.org/wiki/Activation_function)

Are **mathematical equations that determine the output of a neural network**.
The function is attached to each neuron in the network, and determines whether
it should be activated (“fired”) or not, based on whether each neuron’s input is
relevant for the model’s prediction.

Activation functions also help normalize the output of each neuron to a range
between 1 and 0 or between -1 and 1.

```python
import numpy as np
import matplotlib.pyplot as plt

# add default grid lines to the plots
plt.rcParams["axes.grid"] = True
```

### [Logistic Function](https://en.wikipedia.org/wiki/Logistic_function) (Sigmoide Function)

$$ \begin{aligned} \sigma(x) &= \frac{1}{1 + e^{-x}} \\ \sigma'(x) &=
\sigma(x)(1 - \sigma(x)) \\

\end{aligned} $$

$ \forall x \in \mathbb{R} \quad \sigma(x) \in (0, 1) $

```python
# plot sigmoid function
x = np.linspace(-10, 10, 100)
y = 1 / (1 + np.exp(-x))

plt.ylim(-0.1, 1.1)
plt.xlim(-10, 10)
plt.plot(x, y)
```

### Tanh Function

Is similar to the logistic function, but produces an output range between -1 and
1 instead of 0 and 1.

$$
\begin{aligned}
\tanh(x) &= \frac{e^x - e^{-x}}{e^x + e^{-x}} \\
\tanh'(x) &= 1 - \tanh^2(x) \\
\end{aligned}
$$

$ \forall x \in \mathbb{R} \quad \tanh(x) \in (-1, 1) $

```python
# plot tanh function

x = np.linspace(-3, 3, 100)
y = np.tanh(x)

plt.ylim(-1.1, 1.1)
plt.xlim(-3, 3)
plt.plot(x, y)
plot.show()
```

### [Binary Step Function](https://en.wikipedia.org/wiki/Heaviside_step_function)

$$
\begin{aligned}
f(x) &=
    \begin{cases}
        0 & \text{if } x < 0 \\
        1 & \text{if } x \geq 0
    \end{cases} \\
f'(x) &= 0
\end{aligned}
$$

Range: $\{0, 1\}$

```python
# plot sign function
x = np.linspace(-10, 10, 100)
y = np.sign(x)

plt.ylim(-1.1, 1.1)
plt.xlim(-10, 10)
plt.plot(x, y)
```

### [Rectified Linear Unit (ReLU)](https://en.wikipedia.org/wiki/Rectifier_(neural_networks))

ReLU activation function is an activation function defined as the positive part
of its argument:

$$
\begin{aligned}
f(x) &= \max(0, x) \\
f'(x) &=
    \begin{cases}
        0 & \text{if } x < 0 \\
        1 & \text{if } x > 0 \\
        \text{undefined} & \text{if } x = 0
    \end{cases}
\end{aligned}
$$

Range: $[0, \infty)$

```python
x = np.linspace(-10, 10, 100)
y = np.maximum(0, x)
plt.ylim(-1.1, 10)
plt.xlim(-10, 10)
plt.plot(x, y)
plt.show();
```

### Softplus Function

$$
\begin{aligned}
f(x) &= \ln(1 + e^x) \\
f'(x) &= \frac{1}{1 + e^{-x}}
\end{aligned}
$$

Range: $(0, \infty)$

```python
x = np.linspace(-10, 10, 100)
y = np.log(1 + np.exp(x))
plt.ylim(-1.1, 10)
plt.xlim(-10, 10)
plt.plot(x, y)
plt.show();
```

### [Gaussian Function](https://en.wikipedia.org/wiki/Gaussian_function)

$$
\begin{aligned}
f(x) &= e^{-x^2} \\
f'(x) &= -2xe^{-x^2}
\end{aligned}
$$

Range: $(0, 1]$

```python
# plot gaussian function

x = np.linspace(-3, 3, 100)
y = np.exp(-(x**2))

plt.ylim(-0.1, 1.1)
plt.xlim(-3, 3)
plt.plot(x, y)
plot.show()
```

### [Softmax Function](https://en.wikipedia.org/wiki/Softmax_function)

The softmax function is a generalization of the logistic function that
“squashes” a K-dimensional vector $\mathbf{z}$ of arbitrary real values to a
K-dimensional vector $\sigma(\mathbf{z})$ of real values in the range (0, 1)
that add up to 1.

$$
\begin{aligned}
\sigma(\mathbf{z})_j &= \frac{e^{z_j}}{\sum_{k=1}^K e^{z_k}} \\
\sigma(\mathbf{z})_j &= \frac{e^{z_j - \max(\mathbf{z})}}{\sum_{k=1}^K e^{z_k - \max(\mathbf{z})}} \\
\end{aligned}
$$
