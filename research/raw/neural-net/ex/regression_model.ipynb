# Regression Model

```python
import pandas as pd  # data processing, CSV file I/O (e.g. pd.read_csv)
import numpy as np  # linear algebra

from keras.datasets import boston_housing
from keras import models, layers, optimizers, losses, metrics, activations
from keras.callbacks import History
```

```python
(train_data, train_labels), (test_data, test_labels) = boston_housing.load_data()
print(type(train_data), train_data.shape)
print(type((train_data, train_labels)), train_data.shape)
```

```python
train_data.shape
# 404 samples, 13 features
```

Normalizing the data

```python
mean = train_data.mean(axis=0)  # mean of each feature
train_data -= mean  # center the data
std = train_data.std(axis=0)  # standard deviation of each feature
train_data /= std  # normalize the train data

test_data -= mean  # center the test data
test_data /= std  # normalize the test data
```

```python
model: models.Sequential = models.Sequential()

model.add(layers.Dense(64, activation=activations.relu, input_shape=(train_data.shape[1],)))
model.add(layers.Dense(64, activation=activations.relu))
model.add(layers.Dense(1))

model.compile(
    optimizer=optimizers.RMSprop(learning_rate=0.001),  # default lr=0.001
    loss=losses.mse,  # mean squared error (MSE)
    metrics=[metrics.mae]  # mean absolute error (MAE)
)

history: History = model.fit(
    train_data,
    train_labels,
    epochs=100,
    batch_size=1,
    validation_split=0.2
)

history.history.keys()
```
