# [Reduce Overfitting](https://www.tensorflow.org/tutorials/keras/overfit_and_underfit#reduce_overfitting)

### [Regularization](https://towardsdatascience.com/l1-and-l2-regularization-methods-ce25e7fc831c)

Regularization **applies constraints on the complexity of the model** to reduce the overfitting.

- **L1 Regularization** penalizes the model by adding the absolute value of the weights to the loss function.
$$
loss = loss\_funciton + \frac{\lambda}{2} \sum_{i=1}^{n} |w_i|
$$

- **L2 regularization** penalizes the model by adding the square of the weights to the loss function.
$$
loss = loss\_funciton + \frac{\lambda}{2} \sum_{i=1}^{n} w_i^2
$$

Where $\lambda$ is the regularization parameter. The higher the value of $\lambda$, the more the regularization.

### [Dropout](https://towardsdatascience.com/introduction-to-dropout-to-regularize-deep-neural-network-8e9d6b1d4386)

Dropout is a regularization technique that **randomly ignores a subset of the neurons during training**. This forces the network to learn redundant representations of the data. This technique is used to **prevent overfitting**.
