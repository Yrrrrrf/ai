```python
import os  # for file handling
import cv2 as cv  # for image processing
import numpy as np  # for array manipulation
import matplotlib.pyplot as plt  # for plotting
```

```python
IMG_PATH = '.\\..\\resources\\img\\'  # Path to images
directories = [entry for entry in os.listdir(IMG_PATH) if os.path.isdir(os.path.join(IMG_PATH, entry))]  # Filter out directories
```

```python
selected_dir: int = 1  # Select directory to read images from
print(IMG_PATH + f"{directories[selected_dir]}\\")  # print path to selected directory

imgs = [cv.cvtColor(cv.imread(IMG_PATH + f"{directories[selected_dir]}\\{i}"), cv.COLOR_BGR2RGB) for i in os.listdir(IMG_PATH + f"{directories[selected_dir]}\\")]
for img in imgs:
    print(f"{img.shape} {img.dtype}")
```

```python
# Show images
plt.figure(figsize=(20,15))
for i in range(len(imgs)):
    plt.subplot(1, len(imgs), i + 1)
    plt.title(f'Image {i + 1}')
    plt.axis('off')
    plt.imshow(imgs[i])
```

```python
(dummy, output)=cv.Stitcher.create().stitch(imgs) 

match dummy:
	case cv.STITCHER_OK:
		print("stitching ain't successful") 
	case _:
		print('Your Panorama is ready!!!')

plt.figure(figsize=(20,10))
plt.title('Final Panorama')
plt.axis('off')
plt.imshow(output)
plt.show()
```
