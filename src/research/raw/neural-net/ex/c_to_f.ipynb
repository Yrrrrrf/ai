# ANN Celsius to Fahrenheit

This notebook presents a basic Neural Network (NN) to convert Celsius to
Fahrenheit. The NN is trained using the [Keras](https://keras.io/) library with
the [TensorFlow](https://www.tensorflow.org/) backend.

```python
import numpy as np
import matplotlib.pyplot as plt

import tensorflow as tf
```

The NN is trained using a dataset of 1000 samples. The dataset is split into 700
samples for training and 300 samples for testing. The NN is trained for 500
epochs.

```python
def c_to_f(c: float) -> float:
    return c * 1.8 + 32


def f_to_c(f: float) -> float:
    return (f - 32) / 1.8


def c_to_f_array(c_array: np.ndarray) -> np.ndarray:
    return c_array * 1.8 + 32


def f_to_c_array(f_array: np.ndarray) -> np.ndarray:
    return (f_array - 32) / 1.8
```
